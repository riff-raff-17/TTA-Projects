# Session 1 — Python Foundations for Game Programming

## 1. Variables and State

A game's "state" is just its current set of variable values: an `x`/`y`
position, a velocity, a score, a "game over" flag. Nothing exotic — just
values that change over time.

````python
# Game state is just variables
x = 100
y = 50
score = 0
game_over = False

print(f"Player is at ({x}, {y}), score={score}, game_over={game_over}")

# "Updating state" is just reassigning variables
x = x + 5  # move right
score += 10  # scored a point

print(f"After one update: ({x}, {y}), score={score}")
````

````python
# Try it: simulate 3 frames of movement by hand.
# Each "frame", increase x by 2 and y by 1, then print the position.

x, y = 0, 0
for frame in range(3):
    x += 2
    y += 1
    print(f"frame {frame}: ({x}, {y})")
````

## 2. The "Forever Loop" as a Game Loop

A `while` loop that keeps running as long as a condition holds is the
core of any game:

```python
while running:
    # check input, update state, redraw
    ...
```

````python
# A tiny "game loop" with no pygame yet — just printing frames.
running = True
frame_count = 0

while running:
    frame_count += 1
    print(f"Frame {frame_count}: game is running")

    if frame_count >= 5:   # pretend this is "player closed the window"
        running = False

print("Loop ended.")
````

## 3. Conditionals for Branching Behaviour

Most game logic boils down to **numeric comparisons**: was a key pressed?
did a value go out of bounds? do two objects overlap?

````python
position = -5
screen_width = 640

if position < 0:
    print("Off the left edge!")
elif position > screen_width:
    print("Off the right edge!")
else:
    print("On screen.")
````

````python
# Try it: write a check for whether a player's health has hit zero.
health = 0

if health <= 0:
    print("Game over!")
else:
    print(f"Still alive with {health} HP")
````

## 4. Functions for Organising Repeated Actions

Bundling logic into a function (with parameters and a return value) keeps
a growing program readable. Two examples we'll reuse constantly:
`clamp()` (keep a value in a safe range) and a "draw"-style function that
takes the data it needs as arguments.

````python
def clamp(value, low, high):
    """Constrain value to the range [low, high]."""
    if value < low:
        return low
    if value > high:
        return high
    return value


print(clamp(150, 0, 100))  # too high -> 100
print(clamp(-20, 0, 100))  # too low -> 0
print(clamp(42, 0, 100))  # in range -> 42
````

````python
def describe_player(x, y, health):
    """Bundles up a print statement so we don't repeat ourselves."""
    print(f"Player at ({x}, {y}) with {health} HP")


describe_player(10, 20, 100)
describe_player(15, 22, 87)
````

## 5. Lists and Tuples for Coordinates and Collections

- A **tuple** `(x, y)` represents a single 2D point — `pygame` represents
  almost everything (positions, colours, rectangle corners) as tuples.
- A **list** represents a *collection* of things — like a list of body
  segments, or a list of buttons. A growing snake is naturally a growing
  list.

````python
# A point as a tuple
player_pos = (100, 50)
print("x:", player_pos[0], "y:", player_pos[1])

# A collection of points as a list of tuples
body_segments = [(10, 10), (9, 10), (8, 10)]

for segment in body_segments:
    seg_x, seg_y = segment
    print(f"segment at ({seg_x}, {seg_y})")

print("Number of segments:", len(body_segments))
print("Head segment:", body_segments[0])
print("All but the head:", body_segments[1:])
````

````python
# Try it: add a new segment to the FRONT of the list (this is how
# Snake will "move" — insert at front, remove from back).
body_segments.insert(0, (11, 10))
print(body_segments)

body_segments.pop()  # remove the last segment
print(body_segments)
````

## 6. Dictionaries for Structured State

A dictionary groups related values under named keys — useful for
representing one game object's state as a labelled bundle instead of
several loose variables.

````python
ball = {"x": 100, "y": 50, "speed": 4}

print(ball["x"], ball["y"], ball["speed"])

# Updating a dict's values works just like updating a variable
ball["x"] += ball["speed"]
print("After one move:", ball)
````

## 7. Introducing Classes: Bundling Data and Behaviour

A class is a more organised alternative to a dictionary: **data**
(attributes) and the **behaviour** that acts on that data (methods) live
in one place. This is exactly the shape Snake's segments — and much
later, each GUI button — will take.

````python
class Ball:
    def __init__(self, x, y, radius):
        self.x = x
        self.y = y
        self.radius = radius

    def move(self, dx, dy):
        self.x += dx
        self.y += dy


my_ball = Ball(100, 50, 20)
print(my_ball.x, my_ball.y, my_ball.radius)

my_ball.move(5, 0)
print("After move:", my_ball.x, my_ball.y)
````

## 8. Importing and Using External Modules

A module is just a toolbox of pre-written functions someone else wrote,
that you can call instead of rewriting from scratch. We'll use Python's
built-in `random` module today — this directly previews how Snake will
randomly place its food in Session 4.

````python
import random

for _ in range(5):
    print(random.randint(0, 10))
````

## Putting It Together: From Loose Variables to a Class

Here's the same data represented three ways, each one a little more
organised than the last. 

````python
# Step 1: loose variables (the old way)
x = 100
y = 50
radius = 20

print("Step 1:", x, y, radius)
````

````python
# Step 2: grouped in a dictionary
ball = {"x": 100, "y": 50, "radius": 20}

print("Step 2:", ball)
````

````python
# Step 3: grouped in a class (data + behaviour together)
class Ball:
    def __init__(self, x, y, radius):
        self.x = x
        self.y = y
        self.radius = radius

    def move(self, dx, dy):
        self.x += dx
        self.y += dy


my_ball = Ball(100, 50, 20)
my_ball.move(5, 0)
print("Step 3:", my_ball.x, my_ball.y, my_ball.radius)
````
