## 1. Lists

A **list** is an ordered, *mutable* (changeable) collection of items. Think of it like a shopping list where you can add, remove, or reorder items.

Key properties:
- Ordered (items keep their position)
- Mutable (you can change them after creation)
- Allow duplicates
- Can hold mixed types (though usually you'll keep one type per list)

````python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = ["hello", 42, 3.14, True]

print(fruits)
print(numbers)
print(mixed)
````

### Indexing and slicing

Python uses **zero-based indexing** — the first item is at index `0`. Negative indices count from the end.

````python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(fruits[0])  # first item
print(fruits[-1])  # last item
print(fruits[1:3])  # slice: index 1 up to (not including) 3
print(fruits[:2])  # from the start up to index 2
print(fruits[::-1])  # the whole list, reversed
````

### Common list operations

````python
shopping_list = ["milk", "eggs"]

shopping_list.append("bread")  # add to the end
shopping_list.insert(0, "coffee")  # insert at a specific position
shopping_list.remove("eggs")  # remove by value
last_item = shopping_list.pop()  # remove & return the last item
print(shopping_list)
print("Popped:", last_item)

print("Length:", len(shopping_list))
print("Is milk in the list?", "milk" in shopping_list)

shopping_list.sort()  # sort in place
print("Sorted:", shopping_list)
````

### List comprehensions

One of Python's best features; a compact way to build a list from another iterable.

````python
squares = [x**2 for x in range(10)]
print(squares)

evens = [x for x in range(20) if x % 2 == 0]
print(evens)

# You can transform AND filter at the same time
doubled_evens = [x * 2 for x in range(20) if x % 2 == 0]
print(doubled_evens)
````

## 2. Tuples

A **tuple** is just like a list, but **immutable**: once created, it can't be changed. This makes tuples great for data that shouldn't change, like coordinates or fixed records.

Key properties:
- Ordered
- Immutable (cannot be modified after creation)
- Allow duplicates
- Slightly faster and more memory-efficient than lists

````python
# Creating tuples
point = (3, 4)
rgb_color = (255, 0, 128)
single_item_tuple = (42,)  # note the comma! Without it, this is just an int

print(point)
print(rgb_color)
print(type(single_item_tuple))

# Tuples support indexing and slicing just like lists
print(point[0], point[1])
````

````python
# Tuples are immutable - this will raise an error if you uncomment it:
# point[0] = 10   # TypeError: 'tuple' object does not support item assignment

# But you CAN reassign the whole variable
point = (5, 6)
print(point)
````

### Tuple unpacking

This is one of the most useful Python features: assigning multiple variables in one line.

````python
point = (3, 4)
x, y = point
print(f"x = {x}, y = {y}")

# Useful for swapping variables
a, b = 1, 2
a, b = b, a
print(f"a = {a}, b = {b}")

# Useful when looping over pairs
coordinates = [(0, 0), (1, 2), (3, 4)]
for x, y in coordinates:
    print(f"Point at ({x}, {y})")
````

## 3. Dictionaries

A **dictionary** stores data as **key-value pairs**. Instead of looking things up by position (like a list), you look them up by a key, like looking up a word in an actual dictionary.

Key properties:
- Unordered conceptually, but preserves insertion order (Python 3.7+)
- Mutable
- Keys must be unique and immutable (strings, numbers, tuples)
- Extremely fast lookups by key

````python
# Creating dictionaries
person = {"name": "Ada", "age": 36, "occupation": "mathematician"}

print(person)
print(person["name"])  # access by key
print(person.get("age"))  # safer access - won't error if key is missing
print(person.get("email", "no email on file"))  # default value if missing
````

````python
# Modifying dictionaries
person["email"] = "ada@example.com"  # add a new key
person["age"] = 37  # update an existing key
del person["occupation"]  # remove a key

print(person)
print("Keys:", list(person.keys()))
print("Values:", list(person.values()))
print("Items:", list(person.items()))
````

````python
# Looping through a dictionary
inventory = {"apples": 50, "bananas": 30, "cherries": 100}

for item, quantity in inventory.items():
    print(f"We have {quantity} {item}")

# Dictionary comprehensions work too!
doubled_inventory = {item: qty * 2 for item, qty in inventory.items()}
print(doubled_inventory)
````

## 4. Sets

A **set** is an unordered collection of **unique** items. If you've ever used Venn diagrams in math class, sets in Python work the same way, perfect for deduplication and membership testing.

Key properties:
- Unordered
- Mutable
- No duplicates allowed (automatically removed)
- Very fast membership testing (`in` checks)

````python
# Creating sets
fruits_set = {"apple", "banana", "cherry", "apple"}  # duplicate "apple" gets dropped
print(fruits_set)

numbers_list = [1, 2, 2, 3, 3, 3, 4]
unique_numbers = set(numbers_list)  # great trick for deduplication!
print(unique_numbers)
````

````python
# Set operations - just like Venn diagrams
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("Union:", a | b)  # all elements in either set
print("Intersection:", a & b)  # elements in both
print("Difference (a - b):", a - b)  # in a but not b
print("Symmetric difference:", a ^ b)  # in exactly one of the sets

a.add(10)
a.discard(1)
print("Modified a:", a)
````

## 5. Strings (as a Data Structure)

Strings are sequences of characters, and they share a lot in common with lists/tuples: indexing, slicing, and iteration all work. The big difference: **strings are immutable**, just like tuples.

````python
word = "python"

print(word[0])  # first character
print(word[-1])  # last character
print(word[1:4])  # slice
print(word[::-1])  # reverse a string - very handy trick!
print(len(word))

# Strings are immutable
# word[0] = "P"   # this would raise a TypeError

# To "change" a string, you create a new one
capitalized = word.capitalize()
print(capitalized)
````

````python
# Useful string methods
sentence = "  Hello, World!  "

print(repr(sentence.strip()))  # remove leading/trailing whitespace
print(sentence.strip().lower())  # lowercase
print(sentence.strip().upper())  # uppercase
print(sentence.strip().replace("World", "Python"))
print(sentence.strip().split(","))  # split into a list

# Joining a list of strings back together
words = ["Data", "structures", "are", "fun"]
print(" ".join(words))
````

## 7. Bonus Round: The `collections` Module

Python's standard library has a `collections` module with specialized data structures built on top of the basics. Here are three of the most useful ones.

````python
from collections import Counter, defaultdict, namedtuple

# --- Counter: counts hashable items automatically ---
words = "the quick brown fox jumps over the lazy dog the fox runs".split()
word_counts = Counter(words)
print(word_counts)
print(word_counts.most_common(2))  # top 2 most common words

# Remember our manual word-counting exercise earlier? Counter does it in one line!
````

````python
# --- defaultdict: a dict with automatic default values ---
fruit_groups = defaultdict(list)

fruits = ["apple", "avocado", "banana", "blueberry", "cherry"]
for fruit in fruits:
    first_letter = fruit[0]
    fruit_groups[first_letter].append(fruit)  # no need to check if key exists first!

print(dict(fruit_groups))
````

````python
# --- namedtuple: a tuple with named fields, great for lightweight records ---
Point = namedtuple("Point", ["x", "y"])

p1 = Point(3, 4)
print(p1)
print(p1.x, p1.y)  # access by name instead of index!
print(p1[0], p1[1])  # still works like a regular tuple too
````
